# Grammar 3.20

This grammar is well described in the book, in cell below, I will build and compare
states graph and final parsing table.

```python
p = LR0Parser()
p.add_rules(*GRAMMAR_3_20)
p.compute_states_and_edges()
```

```python
print(p.to_tabulate())
```

## Original Parsing Table

LR(0) parsing table for grammar 3.20, example directly from the book.

Indexes of states differ from the book, but I have to check if more or less,
the functionality is the same. That's why I printed out all states and rules
with their corresponding indexes.

#### Update

Except the order, whole logic seems to be the same, so my implementation works! 

```python
for idx, rule in p.indexed_rules.items():
    print(f"{idx:04} {rule}")
```

```python
for idx, state in p.states.items():
    line = f"{idx:04}\t" + ", ".join(str(el) for el in state)
    print(line)
```

```python
GRAPH_PNG_PATH = "static/grammar_3_20_graph.png"
fl = lr_parser_to_mermaid(p, title="grammar 3.20")
mmd.Mermaid(fl).to_png(GRAPH_PNG_PATH)
```
